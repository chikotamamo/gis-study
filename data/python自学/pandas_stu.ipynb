```python
import pandas as pd
df5 = pd.read_csv(path,nrows = 25)
df5['销售金额'] = df5['单价'] * df5['销售数量']
# 分组：
# 3.1 根据 产品类型 进行分组，统计各个类别的 数量  ------> .count
df5.groupby("产品类型")["订单号"].count()
# 3.2 各个类别的销售数量之和
df5.groupby("产品类型")["销售数量"].sum()
# 3.3 各个类别的销售金额之和
df5.groupby("产品类型")["销售金额"].sum()
# 3.4 各个类别的平均单价、最高单价、最低单价
df5.groupby("产品类型")["单价"].agg(['mean','max','min'])
# 3.5 各个类型的销售数量之和、销售金额之和、平均单价
df5.groupby("产品类型").agg({"销售数量":"sum","销售金额":"sum","单价":"mean"})
```
